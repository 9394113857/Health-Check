This code will predict whether the given emails are spam or not. It uses the trained clf model and displays the predictions accordingly.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Load the dataset
data = pd.read_csv('https://raw.githubusercontent.com/9394113857/Data-Sets/9882400f949ec59791995b8be588891f9e0daf48/spam.csv', delimiter=',', encoding='utf-8')

# Add a 'Spam' column
data['Spam'] = data['Category'].apply(lambda x: 1 if x == 'spam' else 0)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(data.Message, data.Spam, test_size=0.25)

# Define the pipeline
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])

# Train the model
clf.fit(X_train, y_train)

# Function to predict spam or not spam based on user input
def predict_spam():
    print("Enter an email message:")
    email = input("Email: ").strip()
    default_response = 'Sounds great! Are you home now?'  # Default placeholder

    if not email or email.lower() == 'yes':
        email = default_response

    prediction = clf.predict([email])
    result = "Spam" if prediction[0] == 1 else "Not Spam"
    print("Prediction:", result)

# Test the function
predict_spam()
```

```python
import joblib

# Save the trained model
joblib.dump(clf, 'spam_detection_model.pkl')

# Load the saved model
loaded_model = joblib.load('spam_detection_model.pkl')

# Function to predict spam or not spam based on user input using the loaded model
def predict_spam_loaded_model():
    print("Enter an email message:")
    email = input("Email: ").strip()
    default_response = 'Sounds great! Are you home now?'  # Default placeholder

    if not email or email.lower() == 'yes':
        email = default_response

    prediction = loaded_model.predict([email])
    result = "Spam" if prediction[0] == 1 else "Not Spam"
    print("Prediction:", result)

# Test the function with the loaded model
predict_spam_loaded_model()
```

```python
import joblib
import os
from rich.console import Console
from rich.table import Table
from rich import box  # Importing box styles
from datetime import datetime
import pytz
import warnings

# Suppress any warnings
warnings.filterwarnings("ignore", category=UserWarning)

# List all files in the current directory and filter out the model files
def list_model_files():
    files = [f for f in os.listdir() if f.endswith('.pkl')]
    return files

# Automatically detect available model file(s)
model_files = list_model_files()

# If no model file is found, exit
if not model_files:
    print("Error: No model files found in the current directory.")
    exit()

# If only one model file is found, use it automatically
if len(model_files) == 1:
    default_model_name = model_files[0]
    print(f"Using default model file: {default_model_name}")

# If multiple model files are found, prompt the user to select one
else:
    print("Multiple model files found:")
    for i, model in enumerate(model_files, start=1):
        print(f"{i}. {model}")

    # Ask the user to select a model
    try:
        choice = int(input(f"\nEnter the number corresponding to the model you want to use (1-{len(model_files)}): "))
        if 1 <= choice <= len(model_files):
            default_model_name = model_files[choice - 1]
            print(f"Using selected model file: {default_model_name}")
        else:
            print("Invalid choice. Exiting.")
            exit()
    except ValueError:
        print("Invalid input. Exiting.")
        exit()

# Try to load the model file
try:
    loaded_model = joblib.load(default_model_name)
except Exception as e:
    print(f"Error loading the model: {e}")
    exit()  # Exit if the model couldn't be loaded

# Function to get current time in IST and formatted date/time string
def get_current_time_ist():
    ist = pytz.timezone('Asia/Kolkata')  # Set IST timezone
    now_ist = datetime.now(ist)  # Get the current time in IST
    formatted_date_time = now_ist.strftime('%Y-%m-%d %H:%M:%S')  # Format the date and time
    return formatted_date_time

# Function to predict spam or not spam based on user input using the loaded model
def predict_spam_loaded_model():
    console = Console()

    # Get current time for displaying
    current_time = get_current_time_ist()

    # Direct and simple email message input prompt with extra line breaks for clarity
    console.print("\n[bold magenta on white] Enter an Email Message to Check Spam Status: [/bold magenta on white]")
    email = input().strip()  # Get user input directly

    # Default response in case of empty input
    default_response = 'Sounds great! Are you home now?'

    # If the input is empty or contains only 'yes', use the default message
    if not email or email.lower() == 'yes':
        email = default_response

    # Add spacing before displaying the user input
    console.print(f"\n[bold cyan]User Input:[/bold cyan] [italic yellow]{email}[/italic yellow]", style="bold")
    console.print("\n" + "-" * 60)  # Separator for better visual structure

    # Display current time of execution in the output
    console.print(f"[bold green]Execution Time: {current_time}[/bold green]")
    console.print("\n" + "-" * 60)  # Separator for better visual structure

    # Make a prediction using the loaded model
    prediction = loaded_model.predict([email])

    # Interpret the result
    result = "Spam" if prediction[0] == 1 else "Not Spam"

    # Add some space before showing prediction result
    console.print(f"\n[bold blue on white]Prediction Result:[/bold blue on white] [bold red]{result}[/bold red]\n")
    console.print("\n" + "-" * 60)  # Separator for better visual structure

    # Create and style the result table with more separation and better alignment
    table = Table(title="[bold green]Spam Detection Result[/bold green]", box=box.ROUNDED, show_header=True)

    # Adding styled columns with clear text alignment and space between them
    table.add_column("Email Message", style="bold white", justify="center", width=40)
    table.add_column("Prediction", style="bold yellow", justify="center", width=20)

    # Add the email message and prediction to the table
    table.add_row(email, result)

    # Print the result in a well-formatted table with extra spacing
    console.print("\n" + "-" * 60)  # Final separator
    console.print(table)

# Test the function with the loaded model
predict_spam_loaded_model()
```

## Interview Preparation Notes

Here are some potential interview questions based on the concepts covered in this notebook, along with simple answers in point format.

### Q1: What is the main goal of this notebook?

### A1: The main goal is to build a model to detect spam emails.

1.  Load a dataset containing emails and their categories (spam or not spam).
2.  Preprocess the text data.
3.  Train a classification model (Naive Bayes) to distinguish between spam and non-spam emails.
4.  Evaluate the model's performance.
5.  Demonstrate how to use the trained model to predict whether a new email is spam.
6.  Save and load the trained model for future use.

### Q2: What dataset was used and where did you get it from?

### A2: The dataset used is "spam.csv".

1.  It was loaded directly from a GitHub repository using a raw URL.
2.  The dataset contains two main columns: 'Category' (indicating if an email is spam or ham/not spam) and 'Message' (the content of the email).

### Q3: What type of machine learning problem is this?

### A3: This is a Supervised Learning problem.

1.  Specifically, it is a Binary Classification task.
2.  We are classifying emails into one of two categories: 'spam' or 'not spam' (ham).

### Q4: How did you prepare the data for the model?

### A4: Data preparation involved a few steps:

1.  A new column named 'Spam' was created.
2.  This column contains numerical labels: 1 for 'spam' and 0 for 'ham'.
3.  The data was split into training and testing sets (75% for training, 25% for testing) using `train_test_split`.

### Q5: What machine learning model was used and why?

### A5: A Multinomial Naive Bayes model was used.

1.  Naive Bayes is a good choice for text classification tasks like this.
2.  Multinomial Naive Bayes is particularly suitable for discrete data, such as word counts in text, which is what the `CountVectorizer` produces.

### Q6: How was the text data converted into a format the model could understand?

### A6: The `CountVectorizer` was used.

1.  It converts a collection of text documents into a matrix of token counts.
2.  Each row represents an email, and each column represents a unique word in the dataset.
3.  The value in each cell is the frequency of that word in that email.

### Q7: Explain the purpose of the `Pipeline` used in the code.

### A7: The `Pipeline` simplifies the workflow.

1.  It chains together multiple processing steps and the final estimator (model) into a single object.
2.  In this notebook, the pipeline combines the `CountVectorizer` (text vectorization) and the `MultinomialNB` model (classification).
3.  This makes the code cleaner and easier to use, especially during training and prediction, as you only need to call `fit` or `predict` on the pipeline object.

### Q8: How did you evaluate the model's performance?

### A8: The model's performance was evaluated using the `score` method.

1.  `clf.score(X_test, y_test)` calculates the accuracy of the model on the test dataset.
2.  Accuracy is the proportion of correctly classified emails (both spam and not spam) out of the total number of emails in the test set.

### Q9: How did you save and load the trained model?

### A9: The `joblib` library was used for saving and loading the model.

1.  `joblib.dump(clf, 'model_filename.pkl')` saves the trained pipeline object to a file.
2.  `joblib.load('model_filename.pkl')` loads the saved model object back into memory.
3.  This allows you to reuse the trained model without retraining it every time.

### Q10: How would you explain the process of predicting a new email's spam status?

### A10: Predicting a new email involves using the trained model.

1.  The new email message is passed to the loaded pipeline's `predict` method.
2.  The pipeline first uses the `CountVectorizer` within it to transform the new email into the numerical format it was trained on.
3.  Then, the Naive Bayes model within the pipeline makes a prediction (0 for not spam, 1 for spam) based on this numerical representation.
4.  The result is interpreted to tell whether the email is predicted as "Spam" or "Not Spam".

### Q11: What are some other potential models you could use for this problem?

### A11: Other models suitable for text classification include:

1.  **Support Vector Machines (SVMs):** Effective for high-dimensional data like text.
2.  **Logistic Regression:** A simple yet powerful linear model for binary classification.
3.  **Tree-based models (like Random Forest or Gradient Boosting):** Can capture non-linear relationships, although they might require more feature engineering for text.
4.  **Deep Learning models (like RNNs, LSTMs, or Transformers):** Can achieve higher accuracy but require more data and computational resources.

### Q12: What are some ways to improve the model's performance?

### A12: Model performance can be improved by:

1.  **Text Preprocessing:** More advanced techniques like removing stopwords, stemming, lemmatization, and handling punctuation/special characters.
2.  **Different Vectorization Techniques:** Using TF-IDF (Term Frequency-Inverse Document Frequency) instead of just Count Vectorizer.
3.  **Hyperparameter Tuning:** Optimizing the parameters of the Naive Bayes model or the vectorizer.
4.  **Using N-grams:** Considering sequences of words (e.g., "free money") instead of just individual words.
5.  **Exploring other models:** Trying the alternative models mentioned in Q11.
6.  **Getting more data:** A larger and more diverse dataset can often improve model generalization.

### Q13: What are some potential challenges with spam detection?

### A13: Challenges in spam detection include:

1.  **Evolving Spam Techniques:** Spammers constantly change their tactics and language to bypass filters.
2.  **Language Variations:** Handling different languages, slang, and misspellings.
3.  **False Positives (Legitimate emails marked as spam):** This is a critical issue, as users can miss important emails.
4.  **False Negatives (Spam emails not detected):** Allows unwanted messages to reach the inbox.
5.  **Handling Attachments and Images:** Spam can be embedded in images or attachments, which text-based models cannot directly analyze.
6.  **Contextual Understanding:** Some emails require understanding the context to determine if they are spam (e.g., a message about a winning lottery from someone you don't know).
