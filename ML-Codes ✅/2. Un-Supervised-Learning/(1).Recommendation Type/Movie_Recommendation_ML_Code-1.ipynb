```python
# Import the CountVectorizer class from the sklearn.feature_extraction.text module
from sklearn.feature_extraction.text import CountVectorizer
```

1.	The CountVectorizer class in scikit-learn is used to convert a collection of text documents into a matrix of token counts.
2.	It tokenizes the text data, builds a vocabulary of known words (tokens), and transforms the text data into a sparse matrix where each row represents a document and each column represents a word in the vocabulary.
3.	Each cell in the matrix contains the count of occurrences of the corresponding word in the document.
4.	This process is useful for preparing text data for machine learning algorithms that require numerical input, such as classification or clustering algorithms.
5.	It allows the algorithms to work with text data by representing it in a format that they can understand and process.



```python
# Create an instance of CountVectorizer
# The 'max_features' parameter sets the maximum number of features (words) to be included in the vocabulary, limiting to 10,000 in this case
# The 'stop_words' parameter is set to 'english' to exclude common English stop words from the vocabulary
cv = CountVectorizer(max_features=10000, stop_words='english')
```

**This code creates an instance of the CountVectorizer class from scikit-learn. Here's a breakdown of the parameters used:**

**1.	max_features:** This parameter specifies the maximum number of features (words) to be included in the vocabulary. In this case, it's set to 10,000, meaning only the top 10,000 most frequent words will be used as features.

**2.	stop_words:** This parameter specifies the list of stop words to be excluded from the vocabulary. By setting it to 'english', common English stop words such as "the", "is", "and", etc., will be ignored during tokenization.

**Overview:**

This instance of CountVectorizer is now configured to tokenize text data, build a vocabulary of up to 10,000 words, and exclude common English stop words.



```python
cv
```

```python
# Use the 'fit_transform' method of CountVectorizer to convert text data into a bag-of-words representation
# The 'new_data['tags'].values.astype('U')' extracts the 'tags' column from 'new_data' and ensures it is of Unicode type
# The resulting sparse matrix is converted to a dense NumPy array using 'toarray()'
vector = cv.fit_transform(new_data['tags'].values.astype('U')).toarray()
```

```python
# Check the shape of the 'vector' array
# The 'shape' attribute provides the dimensions of the array (number of rows and columns)
vector.shape
```

```python
# Import the cosine_similarity function from the sklearn.metrics.pairwise module
from sklearn.metrics.pairwise import cosine_similarity
```

1.	The code imports the cosine_similarity function from the sklearn.metrics.pairwise module.
2.	This function is used to compute the cosine similarity between samples in two matrices.
3.	In the context of natural language processing (NLP) or text analysis, cosine similarity is often used to measure the similarity between two text documents represented as vectors in a high-dimensional space.
4.	It calculates the cosine of the angle between two vectors, providing a measure of similarity regardless of their magnitudes.


```python
# 36 seconds taken
# Calculate the cosine similarity matrix using the 'cosine_similarity' function
# The 'vector' array is used as input, representing the bag-of-words representation of the 'tags' column
# The resulting 'similarity' matrix contains pairwise cosine similarities between rows (documents)
similarity = cosine_similarity(vector)
```

```python
def recommend(movie_title):
    # Find the index of the specified movie in the 'new_data' DataFrame
    index = new_data[new_data['title'] == movie_title].index[0]

    # Calculate the cosine similarity values for the specified movie
    distance = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda vector: vector[1])

    # Print the titles of the top 5 most similar movies
    for i in distance[0:5]:
        print(new_data.iloc[i[0]].title)
```

This function takes a movie title as input, finds its index, calculates cosine similarity values, and then prints the titles of the top 5 most similar movies.

```python
import pandas as pd
import pickle
from IPython.display import display, HTML
import ipywidgets as widgets

# Load the serialized 'new_data' DataFrame using pickle
loaded_movies_list = pickle.load(open('movies_list.pkl', 'rb'))

# Add a serial number (sno) column to the DataFrame
loaded_movies_list['sno'] = range(1, len(loaded_movies_list) + 1)

# Reorganize the columns to have 'sno' as the first column
loaded_movies_list = loaded_movies_list[['sno', 'id', 'title', 'tags']]

# Display the loaded dataset with serial numbers in a tabular format
print("Loaded Dataset:")
display(loaded_movies_list)

# Mock similarity matrix for demonstration purposes
import numpy as np
similarity_matrix = np.random.rand(len(loaded_movies_list), len(loaded_movies_list))

# Function to provide custom recommendations
def custom_recommendation(user_input, movies_data, similarity_matrix):
    try:
        # Find the index of the user input in the DataFrame
        input_index = movies_data[movies_data['title'].str.lower() == user_input.lower()].index[0]

        # Calculate the cosine similarity values for the user input
        distance = sorted(list(enumerate(similarity_matrix[input_index])), reverse=True, key=lambda vector: vector[1])

        # Get the titles of the top 5 most similar movies
        recommendations = []
        for idx, i in enumerate(distance[1:6], 1):  # Exclude the first one as it's the input movie itself
            similar_movie_title = movies_data.iloc[i[0]].title
            recommendations.append({
                'S.No': idx,
                'Recommended Movie': similar_movie_title,
                'Similarity Score': round(i[1], 2)
            })

        # Display recommendations in a table format with borders
        recommendations_df = pd.DataFrame(recommendations)
        display(HTML(f"<h4>Recommendations for: <span style='color: blue;'>{user_input}</span></h4>"))
        display(HTML(recommendations_df.to_html(index=False, border=1, classes='table-bordered')))

    except IndexError:
        print("Movie title not found. Please try a different title.")

# Create an input text box widget for user input
user_input_widget = widgets.Text(
    value='',
    placeholder='Enter a movie title',
    description='Movie Title:',
    disabled=False
)

# Display the input widget
display(user_input_widget)

# Function to handle the input after the user submits
def on_submit(change):
    user_input_title = user_input_widget.value
    custom_recommendation(user_input_title, loaded_movies_list, similarity_matrix)

# Attach the function to the text widget
user_input_widget.on_submit(on_submit)
```

```python
import pickle
from IPython.display import display
from prettytable import PrettyTable

# Load the serialized 'new_data' DataFrame using pickle
loaded_movies_list = pickle.load(open('movies_list.pkl', 'rb'))

# Load the serialized 'similarity' matrix using pickle
similarity_matrix = pickle.load(open('similarity.pkl', 'rb'))

# Add a serial number (sno) column to the DataFrame
loaded_movies_list['sno'] = range(1, len(loaded_movies_list) + 1)

# Reorganize the columns to have 'sno' as the first column
loaded_movies_list = loaded_movies_list[['sno', 'id', 'title', 'tags']]

# Display the loaded dataset with serial numbers in a tabular format
print("Loaded Dataset:")
display(loaded_movies_list)

# Take input from the user
user_input_title = input("\nEnter a movie title for recommendations: ")

# Find the index of the user input in the DataFrame
input_index = loaded_movies_list[loaded_movies_list['title'] == user_input_title].index[0]

# Calculate the cosine similarity values for the user input
distance = sorted(list(enumerate(similarity_matrix[input_index])), reverse=True, key=lambda vector: vector[1])

# Create a PrettyTable for recommendations
recommendations_table = PrettyTable()
recommendations_table.field_names = ["Rank", "Title"]

# Print the titles of the top 5 most similar movies
print(f"\nTop 5 recommendations for '{user_input_title}':")
for rank, i in enumerate(distance[1:6], start=1):  # Exclude the first one as it's the input movie itself
    similar_movie_title = loaded_movies_list.iloc[i[0]].title
    recommendations_table.add_row([rank, similar_movie_title])

# Print recommendations in a table format
print(recommendations_table)
```

**Q: How do you determine if a machine learning problem is supervised, unsupervised, or another type within the AI family?**

**A:** The key factor in determining the type of machine learning problem is the nature of the data you have and the goal you want to achieve:

1.  **Supervised Learning:**
    *   **How to confirm:** You have **labeled data**, meaning your dataset includes both the input features and the corresponding desired output or target variable. The goal is to learn a mapping from inputs to outputs so you can predict the output for new, unseen inputs.
    *   **Subtypes:**
        *   **Classification:** The target variable is categorical (e.g., predicting if an email is spam or not spam, classifying an image as a cat or a dog).
        *   **Regression:** The target variable is continuous or numerical (e.g., predicting house prices, forecasting stock prices).
    *   **How this relates to prediction/insights:** Primarily used for **prediction**.

2.  **Unsupervised Learning:**
    *   **How to confirm:** You have **unlabeled data**, meaning your dataset only contains input features, and there is no predefined target output. The goal is to find hidden patterns, structures, or relationships within the data.
    *   **Subtypes:**
        *   **Clustering:** Grouping similar data points together (e.g., customer segmentation).
        *   **Dimensionality Reduction:** Reducing the number of features while retaining important information (e.g., PCA).
        *   **Association Rule Mining:** Discovering relationships between variables (e.g., "customers who bought item A also bought item B").
    *   **How this relates to prediction/insights:** Primarily used for gaining **insights** and discovering patterns, although the results of unsupervised learning can sometimes be used as features for supervised learning tasks.

3.  **Other AI Family phases (Briefly):**
    *   **Reinforcement Learning:** An agent learns to make decisions by taking actions in an environment and receiving rewards or penalties. This is used for tasks like training game-playing AI or robotics.
    *   **Deep Learning:** A subset of machine learning that uses artificial neural networks with multiple layers to learn complex patterns. Deep learning can be applied to supervised, unsupervised, and reinforcement learning tasks, especially for handling large and complex data like images, audio, and text.
