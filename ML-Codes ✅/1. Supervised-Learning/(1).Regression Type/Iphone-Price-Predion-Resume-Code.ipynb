```python
# Import necessary libraries
import pickle
from rich.console import Console

# Load the pre-trained linear regression model
loaded_linear_regression_model = pickle.load(open('iphone_price_linear_regression.pkl', 'rb'))  # Replace 'iphone_price_linear_regression.pkl' with the actual path to your model file

# Get user input for iphone_version
iphone_version = float(input("Please Enter iphone_version: "))
print()

# Use the model to predict based on the entered iphone_version
prediction_value = loaded_linear_regression_model.predict([[iphone_version]])

# Calculate the amount in rupees
rupees = prediction_value * 79.71

# Create a Console instance for rich text printing
console = Console()

# Print the entered iphone_version in bold green
console.print("The entered iphone_version: [bold green]{}[/bold green]".format(iphone_version))

# Print the predicted value from the model in bold green
console.print("The Dollar Amount is: [bold green]{}[/bold green] Dollars".format(prediction_value[0]))

# Handle potential division by zero when calculating rupees
if prediction_value != 0:
    # Print the converted amount in rupees in bold green
    console.print("The Amount in Rupees is: [bold green]{}[/bold green] Rupees".format(rupees[0]))
else:
    # Print a warning if division by zero occurs
    console.print("[bold red]Warning: Division by zero encountered. Rupees calculation skipped.[/bold red]")

# Print section separators
console.print("======================================")
console.print("======================================")
```

# Points:

**1.	Load Pre-trained Model:**

•	loaded_linear_regression_model = pickle.load(open('iphone_price_linear_regression.pkl', 'rb')): Loads the pre-trained Linear Regression model from the file 'iphone_price_linear_regression.pkl' using Pickle. Replace 'iphone_price_linear_regression.pkl' with the actual path to your model file.

**2.	Get User Input:**

•	iphone_version = float(input("Please Enter iphone_version: ")): Takes user input for the iPhone version and converts it to a float.

**3.	Predict Using the Model:**

•	prediction_value = loaded_linear_regression_model.predict([[iphone_version]]): Uses the loaded model to predict the iPhone price based on the entered version.

**4.	Calculate Amount in Rupees:**

•	rupees = prediction_value * 79.71: Converts the predicted value from dollars to rupees using a fixed conversion rate.

**5.	Rich Text Printing:**

•	console = Console(): Creates a Console instance for rich text printing.

**6.	Print Entered iPhone Version and Predicted Values:**

•	console.print("The entered iphone_version: [bold green]{}[/bold green]".format(iphone_version)): Prints the entered iPhone version in bold green.

•	console.print("The Dollar Amount is: [bold green]{}[/bold green] Dollars".format(prediction_value[0])): Prints the predicted amount in dollars in bold green.

**7.	Handle Division by Zero:**

•	Checks if the predicted value is not zero before calculating the amount in rupees to avoid division by zero.

•	console.print("[bold red]Warning: Division by zero encountered. Rupees calculation skipped.[/bold red]"): Prints a warning message if division by zero occurs.

**8.	Print Section Separators:**

•	console.print("======================================"): Prints section separators for better visual separation.
