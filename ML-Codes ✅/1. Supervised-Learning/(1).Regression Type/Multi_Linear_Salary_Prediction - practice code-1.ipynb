```python
# Importing the libraries
import pandas as pd
```

```python
dataset = pd.read_csv('https://raw.githubusercontent.com/9394113857/Predict-Salary-Analysis/raghu/hiring.csv')
```

```python
dataset.shape
```

```python
dataset.head()
```

```python
# Replacing null values with zero in the 'experience' column
dataset['experience'].fillna(0, inplace=True)

# Replacing null values with the mean of 'test_score' in the 'test_score' column
dataset['test_score'].fillna(dataset['test_score'].mean(), inplace=True)
```

Explanation:
1.	Replace Null Values in 'experience':
•	dataset['experience'].fillna(0, inplace=True): Replaces null values in the 'experience' column with zero. The inplace=True parameter modifies the DataFrame in place.
2.	Replace Null Values in 'test_score':
•	dataset['test_score'].fillna(dataset['test_score'].mean(), inplace=True): Replaces null values in the 'test_score' column with the mean of the non-null values in the same column. The inplace=True parameter modifies the DataFrame in place.



```python
# Extracting the values of all rows and the first three columns from the DataFrame 'dataset' and assigning it to 'X'
X = dataset.iloc[:, :3]
```

**Explanation:**

1.	The code **X = dataset.iloc[:, :3]** extracts all rows and the first three columns of the DataFrame 'dataset' and assigns it to the variable 'X'.
2.	So, 'X' will contain the values of all rows from the first three columns of your dataset, which typically would correspond to your features (such as experience, test_score, interview_score) in the context of a multi-linear regression model.





```python
X
```

```python
# Function to convert words to integer values
def convert_to_int(word):
    word_dict = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5, 'six': 6, 'seven': 7, 'eight': 8,
                'nine': 9, 'ten': 10, 'eleven': 11, 'twelve': 12, 'zero': 0, 0: 0}
    return word_dict[word]
```

Explanation:

•	convert_to_int(word): This function takes a word as input and returns its corresponding integer value based on the predefined dictionary word_dict. The dictionary maps words to their numerical representations. If the input is not found in the dictionary, it returns 0.



```python
# Applying the convert_to_int function to the 'experience' column in DataFrame 'X'
X['experience'] = X['experience'].apply(lambda x: convert_to_int(x))
```

Explanation:

•	X['experience']: Accesses the 'experience' column in DataFrame 'X'.

•	.apply(lambda x: convert_to_int(x)): Applies the convert_to_int function to each element in the 'experience' column using a lambda function. This transforms the words in the 'experience' column to their corresponding integer values.

```python
# Extracting the values of all rows and the last column from the DataFrame 'dataset' and assigning it to 'y'
y = dataset.iloc[:, -1]
```

Explanation:

•	dataset.iloc[:, -1]: Selects all rows (:) and the last column (-1) from the DataFrame 'dataset'.

•	y: Represents the target variable, containing the values of all rows in the last column of the 'dataset'.



```python
y
```

```python
# Importing Linear Regression from scikit-learn
from sklearn.linear_model import LinearRegression

# Creating a Linear Regression model instance
regressor = LinearRegression()
```

Explanation:

1.	Import Linear Regression:

•	from sklearn.linear_model import LinearRegression: Imports the Linear Regression class from the scikit-learn library.

2.	Create Linear Regression Model:

•	regressor = LinearRegression(): Instantiates a Linear Regression model using the imported class. The variable regressor now holds an instance of the Linear Regression model.



```python
# Fitting the Linear Regression model with training data
regressor.fit(X, y)
```

Explanation:

•	regressor.fit(X, y): Fits the Linear Regression model using the training data. The features are represented by the matrix 'X', and the target variable is represented by 'y'. This step trains the model on the provided training dataset.



```python
# Calculate the accuracy score of the Linear Regression model on the training data
regressor.score(X, y)
```

Explanation:

•	regressor.score(X, y): Calculates the R^2 (coefficient of determination) score, which represents the proportion of the variance in the dependent variable ('y') that is predictable from the independent variables ('X'). This method provides the accuracy score of the Linear Regression model on the training data.

```python
# Import the Pickle library
import pickle
```

```python
# Save the trained Linear Regression model to a file using Pickle
pickle.dump(regressor, open('multi_linear_regression_salary_prediction.pkl', 'wb'))
```

Explanation:

pickle.dump(regressor, open('multi_linear_regression_salary_prediction.pkl', 'wb')): Saves the trained Linear Regression model (regressor) to a file named 'multi_linear_regression_salary_prediction.pkl' using Pickle in binary write mode ('wb').







```python
# Load the saved Linear Regression model for comparison
model = pickle.load(open('multi_linear_regression_salary_prediction.pkl', 'rb'))
```

Explanation:

•	model = pickle.load(open('multi_linear_regression_salary_prediction.pkl', 'rb')): Loads the saved Linear Regression model from the file 'multi_linear_regression_salary_prediction.pkl' using Pickle in binary read mode ('rb'). The loaded model is assigned to the variable model.



```python
# The variable 'model' holds the loaded Linear Regression model
model
```
