```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/9394113857/Data-Sets/f82f4f9ad22404bd244417b9ab4e5143a2804fcb/Housing.csv")
```

```python
data.head()
```

```python
print(data.index)
data=data.dropna()
print(data.index)
print(data.columns)
```

```python
data=data.replace(['yes','no','furnished','semi-furnished','unfurnished'],[True,False,1,0.5,0])
print(data.head())
#splitting into input and output labels
y=data['price']
print(y.head())
x=data.drop(['price'],axis=1)
print(x.columns)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
results=model.predict(x_test)
R2_score=r2_score(y_test,results)
print(R2_score)
```
